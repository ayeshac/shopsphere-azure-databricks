```python
base_path = "abfss://shopsphere@stshopsphere2026.dfs.core.windows.net/raw"

customers_path = f"{base_path}/customers/customers.csv"
products_path = f"{base_path}/products/products.csv"
orders_path = f"{base_path}/orders/orders.csv"
```

```python
customers_df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(customers_path)
)

products_df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(products_path)
)

orders_df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(orders_path)
)
```

```python
print(f"Customers: {customers_df.count():,}")
print(f"Products:  {products_df.count():,}")
print(f"Orders:    {orders_df.count():,}")
```

```python
bronze_base = "abfss://shopsphere@stshopsphere2026.dfs.core.windows.net/bronze"

customers_bronze_path = f"{bronze_base}/customers"

(customers_df.write
    .format("delta")
    .mode("overwrite")
    .save(customers_bronze_path)
)
```

```python
spark.read.format("delta").load(customers_bronze_path).display()
```

```python
products_bronze_path = f"{bronze_base}/products"

(
    products_df.write
    .format("delta")
    .mode("overwrite")
    .save(products_bronze_path)
)
```

```python
orders_bronze_path = f"{bronze_base}/orders"

(
    orders_df.write
    .format("delta")
    .mode("overwrite")
    .save(orders_bronze_path)
)
```

```python
spark.read.format("delta").load(products_bronze_path).display()
```

```python
print("Customers:", spark.read.format("delta").load(customers_bronze_path).count())
print("Products :", spark.read.format("delta").load(products_bronze_path).count())
print("Orders   :", spark.read.format("delta").load(orders_bronze_path).count())
```
